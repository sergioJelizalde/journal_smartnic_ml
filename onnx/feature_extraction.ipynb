```python
import os
import pandas as pd
from glob import glob

# ─── Paths ─────────────────────────────────────────────────────────────────────
recon_folder = r"D:\datasets\pcaps_cic\recon"
output_folder = recon_folder  # Save in same folder
N_values = [4, 8, 16, 32, 64, 128]

# ─── Flow‐key generator ────────────────────────────────────────────────────────
def get_flow_key(row):
    proto = int(row['proto'])
    src_ip = row['src_ip']
    dst_ip = row['dst_ip']
    if proto == 6:        # TCP
        src_port = int(row['tcp_src'])
        dst_port = int(row['tcp_dst'])
    elif proto == 17:     # UDP
        src_port = int(row['udp_src'])
        dst_port = int(row['udp_dst'])
    else:                 # ICMP or other
        src_port = dst_port = -1
    return (src_ip, dst_ip, src_port, dst_port, proto)

# ─── Accumulator ───────────────────────────────────────────────────────────────
recon_records = []

# ─── Process Each Recon CSV ─────────────────────────────────────────────────────
csv_files = glob(os.path.join(recon_folder, "*.csv"))
print(f"Found {len(csv_files)} Recon CSV files.")

for csv_file in csv_files:
    print(f"Processing {os.path.basename(csv_file)}...")
    df = pd.read_csv(csv_file, low_memory=False)

    df = df.dropna(subset=["timestamp", "src_ip", "dst_ip", "proto", "frame_len", "ip_len"])
    df["timestamp"] = pd.to_numeric(df["timestamp"], errors="coerce")
    df["frame_len"] = pd.to_numeric(df["frame_len"], errors="coerce")
    df["ip_len"] = pd.to_numeric(df["ip_len"], errors="coerce")
    df = df.dropna(subset=["timestamp", "frame_len", "ip_len"])

    for col in ["tcp_src", "tcp_dst", "udp_src", "udp_dst"]:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(-1).astype(int)

    df["flow_key"] = df.apply(get_flow_key, axis=1)
    df = df.sort_values("timestamp")
    df["label"] = "Recon"

    for flow_key, flow_df in df.groupby("flow_key"):
        flow_df = flow_df.copy()
        flow_df["iat"] = flow_df["timestamp"].diff().fillna(0)
        for N in N_values:
            if len(flow_df) < N:
                continue
            segment = flow_df.iloc[:N]
            src_ip, dst_ip, src_port, dst_port, proto = flow_key

            rec = {
                "src_ip":         src_ip,
                "dst_ip":         dst_ip,
                "src_port":       src_port,
                "dst_port":       dst_port,
                "proto":          proto,
                "label":          "Recon",
                "N":              N,
                "duration":       segment["timestamp"].iloc[-1] - segment["timestamp"].iloc[0],
                "iat_mean":       segment["iat"].mean(),
                "iat_min":        segment["iat"].min(),
                "iat_max":        segment["iat"].max(),
                "frame_len_mean": segment["frame_len"].mean(),
                "frame_len_min":  segment["frame_len"].min(),
                "frame_len_max":  segment["frame_len"].max(),
                "ip_len_mean":    segment["ip_len"].mean(),
                "ip_len_min":     segment["ip_len"].min(),
                "ip_len_max":     segment["ip_len"].max(),
                "tcp_flag_count": segment["tcp_flags"].count(),
                "bytes_sent":     segment["ip_len"].sum(),
            }

            recon_records.append(rec)

# ─── Save ───────────────────────────────────────────────────────────────────────
if recon_records:
    out_path = os.path.join(output_folder, "Recon_flows.csv")
    pd.DataFrame(recon_records).to_csv(out_path, index=False)
    print(f"Saved Recon_flows.csv to: {out_path}")
else:
    print("No valid flow segments extracted.")
```

```python
import os
import pandas as pd
from glob import glob

# ─── Configuration ────────────────────────────────────────────────────────────
data_folder = r"D:\datasets\pcaps_cic"
output_folder = os.path.join(data_folder, "flow_features")
os.makedirs(output_folder, exist_ok=True)
N_values = [4, 8, 16, 32, 64, 128]

# ─── Flow‐key function (TCP, UDP, ICMP) ─────────────────────────────────────
def get_flow_key(row):
    proto = int(row['proto'])
    src_ip = row['src_ip']
    dst_ip = row['dst_ip']
    if proto == 6:        # TCP
        src_port = int(row['tcp_src'])
        dst_port = int(row['tcp_dst'])
    elif proto == 17:     # UDP
        src_port = int(row['udp_src'])
        dst_port = int(row['udp_dst'])
    else:                 # ICMP or other (no ports)
        src_port = dst_port = -1
    return (src_ip, dst_ip, src_port, dst_port, proto)

# ─── Accumulators ─────────────────────────────────────────────────────────────
benign_records = []
ddos_records   = []

# ─── Main Loop: process each CSV ───────────────────────────────────────────
csv_files = glob(os.path.join(data_folder, "*.csv"))
print(f"Found {len(csv_files)} CSV files.")

for csv_file in csv_files:
    print(f"Processing {os.path.basename(csv_file)}...")
    df = pd.read_csv(csv_file, low_memory=False)

    # 1) Drop rows missing strictly‐required fields
    df = df.dropna(subset=["timestamp", "src_ip", "dst_ip", "proto", "frame_len", "ip_len"])

    # 2) Convert those to numeric and drop any that failed
    df["timestamp"] = pd.to_numeric(df["timestamp"], errors="coerce")
    df["frame_len"]  = pd.to_numeric(df["frame_len"],  errors="coerce")
    df["ip_len"]     = pd.to_numeric(df["ip_len"],     errors="coerce")
    df = df.dropna(subset=["timestamp", "frame_len", "ip_len"])

    # 3) Parse ports, filling NaN with -1
    for col in ["tcp_src", "tcp_dst", "udp_src", "udp_dst"]:
        df[col] = pd.to_numeric(df[col], errors="coerce").fillna(-1).astype(int)

    # 4) Create directional 5-tuple key and sort by time
    df["flow_key"] = df.apply(get_flow_key, axis=1)
    df = df.sort_values("timestamp")

    # 5) Label from filename
    label = "DDoS" if "ddos" in os.path.basename(csv_file).lower() else "Benign"
    df["label"] = label

    # 6) Group and compute per‐flow stats
    for flow_key, flow_df in df.groupby("flow_key"):
        flow_df = flow_df.copy()
        flow_df["iat"] = flow_df["timestamp"].diff().fillna(0)
        for N in N_values:
            if len(flow_df) < N:
                continue
            segment = flow_df.iloc[:N]
            src_ip, dst_ip, src_port, dst_port, proto = flow_key

            rec = {
                "src_ip":          src_ip,
                "dst_ip":          dst_ip,
                "src_port":        src_port,
                "dst_port":        dst_port,
                "proto":           proto,
                "label":           label,
                "N":               N,
                "duration":        segment["timestamp"].iloc[-1] - segment["timestamp"].iloc[0],
                "iat_mean":        segment["iat"].mean(),
                "iat_min":         segment["iat"].min(),
                "iat_max":         segment["iat"].max(),
                "frame_len_mean":  segment["frame_len"].mean(),
                "frame_len_min":   segment["frame_len"].min(),
                "frame_len_max":   segment["frame_len"].max(),
                "ip_len_mean":     segment["ip_len"].mean(),
                "ip_len_min":      segment["ip_len"].min(),
                "ip_len_max":      segment["ip_len"].max(),
                "tcp_flag_count":  segment["tcp_flags"].count(),
                "bytes_sent":      segment["ip_len"].sum(),
            }

            if label == "Benign":
                benign_records.append(rec)
            else:
                ddos_records.append(rec)

# ─── Save Results ────────────────────────────────────────────────────
if benign_records:
    pd.DataFrame(benign_records).to_csv(
        os.path.join(output_folder, "Benign_flows.csv"), index=False)
    print("Saved Benign_flows.csv")

if ddos_records:
    pd.DataFrame(ddos_records).to_csv(
        os.path.join(output_folder, "DDoS_flows.csv"), index=False)
    print("Saved DDoS_flows.csv")

print(" All flows processed.")  # Done!
```

```python
import os
import time
import socket
import numpy as np
import pandas as pd
from glob import glob
from scapy.all import rdpcap, IP, TCP, UDP

# ─── SimpleFlow and helpers ───────────────────────────────────────────────────
class SimpleFlow:
    def __init__(self, proto, ip1, ip2, port1, port2):
        self.proto = proto
        self.timestamps = []
        self.pkt_sizes = []
        self.bytes_sent = 0
        self.bytes_received = 0

    def add_packet(self, ts, size, direction):
        self.timestamps.append(float(ts))
        self.pkt_sizes.append(size)
        if direction == "fwd":
            self.bytes_sent += size
        else:
            self.bytes_received += size

    def is_ready(self, N):
        return len(self.pkt_sizes) >= N

    def compute_features(self, label):
        iats = np.diff(self.timestamps)
        return {
            "protocol":       self.proto,
            "mean_iat":       np.mean(iats) if len(iats)>0 else 0,
            "min_iat":        np.min(iats)  if len(iats)>0 else 0,
            "max_iat":        np.max(iats)  if len(iats)>0 else 0,
            "mean_pkt_size":  np.mean(self.pkt_sizes),
            "min_pkt_size":   np.min(self.pkt_sizes),
            "max_pkt_size":   np.max(self.pkt_sizes),
            "bytes_sent":     self.bytes_sent,
            "bytes_received": self.bytes_received,
            "packet_count":   len(self.pkt_sizes),
            "label":          label
        }

def normalize_flow_key(src_ip, dst_ip, src_port, dst_port, proto):
    key1 = (src_ip, dst_ip, src_port, dst_port)
    key2 = (dst_ip, src_ip, dst_port, src_port)
    if key1 <= key2:
        return (*key1, proto), "fwd"
    else:
        return (*key2, proto), "rev"

# ─── Flow Extraction ──────────────────────────────────────────────────────────
def process_pcap(pcap_path, label, N=10):
    print(f"\nProcessing {os.path.basename(pcap_path)} as {label}")
    packets = rdpcap(pcap_path)
    flows = {}
    features = []

    for pkt in packets:
        if IP not in pkt:
            continue
        ip = pkt[IP]
        proto = ip.proto
        src_ip, dst_ip = ip.src, ip.dst

        if proto == 6 and TCP in pkt:
            src_port, dst_port = pkt[TCP].sport, pkt[TCP].dport
        elif proto == 17 and UDP in pkt:
            src_port, dst_port = pkt[UDP].sport, pkt[UDP].dport
        else:
            src_port = dst_port = 0

        key, direction = normalize_flow_key(src_ip, dst_ip, src_port, dst_port, proto)
        if key not in flows:
            flows[key] = SimpleFlow(proto, *key[:4])
        flows[key].add_packet(pkt.time, len(pkt), direction)

        if flows[key].is_ready(N):
            features.append(flows[key].compute_features(label))
            del flows[key]

    return pd.DataFrame(features)

# ─── Main: list DDoS PCAPs, choose one, process ──────────────────────────────
if __name__ == "__main__":
    pcap_folder = r"D:\datasets\pcaps_cic"  # Update if needed
    N = 10

    # List all DDoS PCAP/PCAPNG files
    all_pcaps = glob(os.path.join(pcap_folder, "*.pcap")) + \
                glob(os.path.join(pcap_folder, "*.pcapng"))
    ddos_pcaps = [f for f in all_pcaps if "ddos" in os.path.basename(f).lower()]

    if not ddos_pcaps:
        print("No DDoS PCAP files found in", pcap_folder)
        exit()

    print("Available DDoS PCAPs:")
    for i, path in enumerate(ddos_pcaps):
        print(f" [{i}] {os.path.basename(path)}")

    choice = int(input("\nEnter the index of the PCAP to process: "))
    sel_path = ddos_pcaps[choice]
    label = "DDoS"

    start = time.time()
    df_out = process_pcap(sel_path, label, N=N)
    elapsed = time.time() - start

    print(f"\nExtracted {len(df_out)} flows in {elapsed:.2f}s")
    print(df_out.head())

    # Save results
    out_csv = os.path.join(pcap_folder,
                f"{os.path.splitext(os.path.basename(sel_path))[0]}_flows_N{N}.csv")
    df_out.to_csv(out_csv, index=False)
    print("Saved features to", out_csv)

```
